```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read file
df = pd.read_csv("smart_manufacturing_data.csv")
df.head()
```

```python
# Shape of dataset
df.shape
```

```python
# Information of dataset
df.info()
```

```python
# Convert timestamp to datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
# Statistical summary
df.describe()
```

```python
# Check if any duplicate values
df.duplicated().sum()
```

```python
# Column names
df.columns
```

```python
# Histogram
plt.hist(df["temperature"], bins=20)
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.title("Temperature Distribution")
plt.show()
```

```python
# Histogram
plt.hist(df["vibration"], bins=20)
plt.xlabel("Vibration")
plt.ylabel("Frequency")
plt.title("Vibration Distribution")
plt.show()
```

```python
# Box Plot
plt.boxplot(df["temperature"])
plt.title("Temperature boxplot")
plt.ylabel("Temperature")
plt.show()
```

```python
# Box Plot
plt.boxplot(df["vibration"])
plt.title("Vibration boxplot")
plt.ylabel("Vibration")
plt.show()
```

```python
# Boxplot
plt.boxplot(df["energy_consumption"])
plt.title("Energy Consumption Boxplot")
plt.ylabel("Energy Consumption")
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(10, 8))
corr = df.corr(numeric_only=True)
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Matrix")
plt.xticks(rotation=45)
plt.show()
```

```python
# Bar chart
df["failure_type"].value_counts().plot(kind="bar")
plt.title("Failure Type Distribution")
plt.xlabel("Failure Type")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.show()
```

```python
# Bar Chart
df["machine_status"].value_counts().plot(kind="bar")
plt.title("Machine Status Distribution")
plt.xlabel("Machine Status")
plt.ylabel("Count")
plt.show()
```

```python
# Bar Chart
df["anomaly_flag"].value_counts().plot(kind="bar")
plt.title("Anomaly Distribution")
plt.xlabel("Anomaly Flag")
plt.ylabel("Count")
plt.show()
```

```python
# Calculate daily average temperature
daily_temp = df.groupby(df["timestamp"].dt.date)["temperature"].mean()
# Line chart
plt.plot(daily_temp.index, daily_temp.values)
plt.title("Daily Average Temperature Trend")
plt.xlabel("Date")
plt.ylabel("Average Temperature")
plt.xticks(rotation=45)
plt.show()
```
